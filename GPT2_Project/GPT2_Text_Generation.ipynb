```python
from transformers import GPT2Tokenizer, GPT2LMHeadModel

print("Import Successful")
```

```python
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
model = GPT2LMHeadModel.from_pretrained("gpt2")

print("GPT-2 Model Loaded Successfully!")
```

```python
text = "Artificial Intelligence is transforming education because"

inputs = tokenizer(text, return_tensors="pt")

output = model.generate(
    **inputs,
    max_new_tokens=50,
    do_sample=True,
    temperature=0.8,
    top_k=50,
    top_p=0.95
)

generated_text = tokenizer.decode(output[0], skip_special_tokens=True)
print(generated_text)
```

```python
import pandas as pd

experiments = [
    "Healthcare",
    "Education",
    "NLP",
    "Banking",
    "Deep Learning"
]

outputs = [
    "Machine Learning is used in healthcare because it provides a great opportunity for clinicians to understand patients' needs and outcomes.",
    "Artificial Intelligence is transforming education because it allows students to make and understand knowledge.",
    "Natural Language Processing helps computers to interpret, categorize, and analyze data in natural language.",
    "Artificial Intelligence is used in banking because it is highly efficient and helps in fast transactions.",
    "Deep Learning is important because it helps improve Artificial Intelligence and solve real-world problems."
]

lengths = [len(output.split()) for output in outputs]

df = pd.DataFrame({
    "Experiment": experiments,
    "Output Length": lengths
})

print(df)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.bar(df["Experiment"], df["Output Length"])

plt.title("GPT-2 Output Length Comparison")
plt.xlabel("Experiments")
plt.ylabel("Number of Words")

plt.show()
```
